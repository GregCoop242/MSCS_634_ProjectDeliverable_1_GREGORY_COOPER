```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('archive/flights.csv')
df.head()
```

```python
#inspecrt the data

df.shape
df.info()
df.describe()
```

```python
#Handle Missing Values

#Detect the values first

df.isnull().sum()     # count per column
df.isnull().mean() * 100   #percentage per column
```

```python
#drop rows with a missing value or fill in missing value

#df.dropna()
df.dropna(thresh=10)
#df.dropna(axis=1, how='all')  #drop call columns that are entirely NaN
```

```python
#remove any duplicates

df.duplicated().sum()  # count duplicate rows

#no duplicates: np.int64(0)
```

```python
# Scatter plot to spot individual outlier points
plt.figure(figsize=(8, 4))
#plt.scatter(range(len(df)), df['ARRIVAL_DELAY'], alpha=0.5, color='steelblue')
plt.scatter(df['DAY_OF_WEEK'], df['ARRIVAL_DELAY'], alpha=0.5, color='steelblue')

plt.title('ARRIVAL DELAY - Scatter Plot')
plt.xlabel('DAY_OF_WEEK')
plt.ylabel('ARRIVAL_DELAY')
plt.tight_layout()
plt.show()
```

```python
#Detect Outliers with IQR

Q1 = df['ARRIVAL_DELAY'].quantile(0.25)
Q3 = df['ARRIVAL_DELAY'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['ARRIVAL_DELAY'] < lower) | (df['ARRIVAL_DELAY'] > upper)]
print(f"Outliers found: {len(outliers)}")
```

```python
df = df[(df['ARRIVAL_DELAY'] >= lower) & (df['ARRIVAL_DELAY'] <= upper)]
```

```python
# Scatter plot to spot individual outlier points
plt.figure(figsize=(8, 4))
#plt.scatter(range(len(df)), df['ARRIVAL_DELAY'], alpha=0.5, color='steelblue')
plt.scatter(df['DAY_OF_WEEK'], df['ARRIVAL_DELAY'], alpha=0.5, color='steelblue')

plt.title('ARRIVAL DELAY - Scatter Plot')
plt.xlabel('DAY_OF_WEEK')
plt.ylabel('ARRIVAL_DELAY')
plt.tight_layout()
plt.show()
```
